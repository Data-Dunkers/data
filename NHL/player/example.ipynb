# NHL Player Example

This notebook loads the NHL player CSV and charts the top scorers.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(r"https://github.com/Data-Dunkers/data/blob/main/NHL/player/nhl_player_stats_2025-2026.csv?raw=1")
df = df.sort_values("PTS", ascending=False).head(20)

fig = px.bar(
    df,
    x="Name",
    y="PTS",
    color="Team",
    title="NHL Player Stats: Top 20 Players by Points",
    labels={"Name": "Player", "PTS": "Points"},
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

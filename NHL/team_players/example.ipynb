# NHL Team Players Example

This notebook loads one team-season file and makes a simple bar chart of the top players by points.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(
    r"https://github.com/Data-Dunkers/data/blob/main/data/NHL/team_players/2025-2026/NYR_2025-2026_players.csv?raw=1"
)

df = df.sort_values("PTS", ascending=False).head(10)
fig = px.bar(df, x="Name", y="PTS", title="New York Rangers Top 10 Players by Points")
fig.show()
```

# NHL Standings Example

This notebook loads the NHL standings CSV and makes a simple bar chart of team points.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(r"https://github.com/Data-Dunkers/data/blob/main/data/NHL/standings/nhl_standings_now.csv?raw=1")
df = df.sort_values("points", ascending=False)

fig = px.bar(
    df,
    x="teamName_default",
    y="points",
    title="NHL Standings: Points by Team",
    labels={"teamName_default": "Team", "points": "Points"},
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

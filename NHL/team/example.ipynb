# NHL Team Example

This notebook loads the NHL team summary CSV and charts skater points by team.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(r"https://github.com/Data-Dunkers/data/blob/main/data/NHL/team/nhl_team_club_stats_now.csv?raw=1")
df = df.sort_values("SkaterPTS", ascending=False)

fig = px.bar(
    df,
    x="Team",
    y="SkaterPTS",
    title="NHL Team Club Stats: Skater Points by Team",
    labels={"Team": "Team", "SkaterPTS": "Skater Points"},
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```
